# Regresión simbólica de KF — EJECUCION_1

Este cuaderno prueba el flujo con la hoja `EJECUCION_1` del archivo `1356_VARIACIONES_1.xlsx`. La hoja contiene 1.014 casos; `RUN_2` (1.356 casos) queda reservado para una ejecución posterior independiente.

La respuesta se calcula con la ecuación confirmada:

$$KF = \frac{12(1-\nu^2)\,FCR\,h_w}{\pi^2 E t_w^3}$$

## Variables del modelo

- `SS_HW = ss/hw`: separación longitudinal relativa entre perforaciones y altura del alma.
- `A_R = a/R`: parámetro adimensional de curvatura.
- `DO_HW = do/hw`: diámetro relativo de las perforaciones circulares.
- `HW_TW = hw/tw`: esbeltez del alma.
- `N_PERF`: número de perforaciones (variable discreta; en este libro se codifica como 2 o 4).
- `KF`: coeficiente de pandeo elástico derivado de `FCR_24`.

Se usan $E=210000\;\mathrm{N/mm^2}$, $\nu=0.30$, `FCR_24` en N y longitudes en mm. El ajuste usa las variables adimensionales `ss_hw`, `a_R`, `Do_hw`, `hw_tw` y `n_perf`. No se escriben resultados en el Excel.

```python
from pathlib import Path
import os
import math
import tempfile

os.environ.setdefault('JULIA_PYTHONCALL_EXE', '/usr/bin/python3.12')

import numpy as np
import pandas as pd
import sympy as sp
from sklearn.model_selection import train_test_split
from sklearn.metrics import mean_absolute_error, mean_squared_error, r2_score
from pysr import PySRRegressor

WORKBOOK = Path('1356_VARIACIONES_1.xlsx')
SHEET = 'EJECUCION_1'
OUTPUT_DIR = Path('salidas_regresion_simbolica') / 'ejecucion1_sr'
E_N_PER_MM2 = 210000.0
NU = 0.30
FEATURES = ['ss_hw', 'a_R', 'Do_hw', 'hw_tw', 'n_perf']
REQUIRED = FEATURES + ['hw', 'tw', 'FCR_24']

if not WORKBOOK.is_file():
    raise FileNotFoundError(f'No se encontró el libro: {WORKBOOK.resolve()}')

print('PySR listo. Libro:', WORKBOOK.resolve())
```

```python
df = pd.read_excel(WORKBOOK, sheet_name=SHEET)
missing = sorted(set(REQUIRED) - set(df.columns))
if missing:
    raise ValueError(f'Faltan columnas requeridas en {SHEET}: {missing}')
if len(df) != 1014:
    raise ValueError(f'Se esperaban 1.014 casos para la prueba; se encontraron {len(df)}.')
if df[REQUIRED].isna().any().any():
    raise ValueError('Hay valores vacíos en las variables o en los datos físicos requeridos.')
if not np.isfinite(df[REQUIRED].to_numpy(dtype=float)).all():
    raise ValueError('Hay valores no finitos en las variables o en los datos físicos requeridos.')
if (df[['hw', 'tw', 'FCR_24']] <= 0).any().any():
    raise ValueError('hw, tw y FCR_24 deben ser positivos para calcular KF.')
if not np.isclose(df['n_perf'], np.round(df['n_perf'])).all():
    raise ValueError('n_perf debe contener números enteros.')

df['n_perf'] = df['n_perf'].astype(int)
df['KF'] = (
    12 * (1 - NU**2) * df['FCR_24'] * df['hw']
    / (math.pi**2 * E_N_PER_MM2 * df['tw']**3)
)

X = df[FEATURES].to_numpy(dtype=np.float64)
y = df['KF'].to_numpy(dtype=np.float64)
if not np.isfinite(y).all() or (y <= 0).any():
    raise ValueError('KF contiene valores no positivos o no finitos.')

print(f'Casos: {len(df)}')
print('Rango KF:', float(y.min()), 'a', float(y.max()))
display(df[FEATURES + ['KF']].head())
```

```python
X_train, X_test, y_train, y_test = train_test_split(
    X, y, test_size=0.20, random_state=42
)
print(f'Entrenamiento: {len(y_train)} casos; prueba: {len(y_test)} casos.')
```

```python
# Comparación de configuraciones: selección en validación interna; el conjunto test queda aislado.
from scipy.stats import pearsonr, spearmanr
from sklearn.model_selection import train_test_split

X_fit, X_val, y_fit, y_val = train_test_split(
    X_train, y_train, test_size=0.25, random_state=73
)
search_configs = [
    {'name': 'balanced', 'niterations': 120, 'populations': 12, 'population_size': 32,
     'maxsize': 20, 'parsimony': 0.003, 'weight_add_node': 0.25,
     'weight_insert_node': 0.20, 'weight_delete_node': 0.12,
     'weight_mutate_constant': 0.18, 'weight_mutate_operator': 0.08,
     'weight_mutate_feature': 0.12},
    {'name': 'exploration', 'niterations': 160, 'populations': 16, 'population_size': 36,
     'maxsize': 22, 'parsimony': 0.0005, 'weight_add_node': 0.28,
     'weight_insert_node': 0.22, 'weight_delete_node': 0.10,
     'weight_mutate_constant': 0.16, 'weight_mutate_operator': 0.10,
     'weight_mutate_feature': 0.14},
    {'name': 'compacta', 'niterations': 120, 'populations': 12, 'population_size': 32,
     'maxsize': 16, 'parsimony': 0.01, 'weight_add_node': 0.20,
     'weight_insert_node': 0.15, 'weight_delete_node': 0.20,
     'weight_mutate_constant': 0.20, 'weight_mutate_operator': 0.08,
     'weight_mutate_feature': 0.12},
]

def fit_symbolic(config, x, target, run_id):
    params = {k: v for k, v in config.items() if k != 'name'}
    with tempfile.TemporaryDirectory(prefix=f'pysr_{run_id}_') as work_dir:
        estimator = PySRRegressor(
            **params,
            binary_operators=['+', '-', '*', '/'],
            unary_operators=['square'],
            model_selection='best',
            random_state=42,
            deterministic=True,
            parallelism='serial',
            timeout_in_seconds=900,
            verbosity=0,
            progress=False,
            output_directory=work_dir,
            run_id=run_id,
        )
        estimator.fit(x, target)
    return estimator

def score_predictions(observed, predicted):
    residual = predicted - observed
    pearson = pearsonr(observed, predicted).statistic if np.std(predicted) > 0 else np.nan
    spearman = spearmanr(observed, predicted).statistic if np.std(predicted) > 0 else np.nan
    return {
        'MAE': mean_absolute_error(observed, predicted),
        'RMSE': float(np.sqrt(mean_squared_error(observed, predicted))),
        'R2': r2_score(observed, predicted),
        'MAPE_pct': float(np.mean(np.abs(residual / observed)) * 100),
        'bias_pred_minus_obs': float(np.mean(residual)),
        'residual_std': float(np.std(residual, ddof=1)),
        'pearson_r': float(pearson),
        'spearman_rho': float(spearman),
        'positive_prediction_rate': float(np.mean(predicted > 0)),
    }

validation_rows = []
for config in search_configs:
    candidate = fit_symbolic(config, X_fit, y_fit, f'e1_{config["name"]}')
    candidate_prediction = candidate.predict(X_val)
    row = {'configuration': config['name'], **score_predictions(y_val, candidate_prediction)}
    row['expression'] = str(candidate.sympy())
    validation_rows.append(row)
    print(f"{config['name']}: RMSE={row['RMSE']:.3f}, MAE={row['MAE']:.3f}, R²={row['R2']:.4f}")

validation_metrics = pd.DataFrame(validation_rows).sort_values(['RMSE', 'MAE'])
display(validation_metrics[['configuration', 'MAE', 'RMSE', 'R2', 'MAPE_pct', 'expression']])
best_name = validation_metrics.iloc[0]['configuration']
best_config = next(c for c in search_configs if c['name'] == best_name)
print('Configuración seleccionada solo con validación:', best_name)

# Refit con todo el 80 % de entrenamiento; test se utiliza una sola vez para evaluación final.
model = fit_symbolic(best_config, X_train, y_train, 'e1_final_test')
y_pred = model.predict(X_test)
best_equation = model.sympy()
test_metrics = score_predictions(y_test, y_pred)

print('Ecuación simbólica (x0=SS_HW, x1=A_R, x2=DO_HW, x3=HW_TW, x4=N_PERF):')
print(best_equation)
feature_symbols = {symbol: sp.Symbol(FEATURES[int(str(symbol)[1:])].upper())
                   for symbol in best_equation.free_symbols}
named_equation = best_equation.xreplace(feature_symbols)
print('Expresión con nombres de ingeniería:', named_equation)
print('Métricas en test:', test_metrics)

# Intervalo bootstrap del MAE en test; cuantifica variabilidad muestral, no incertidumbre física.
bootstrap_rng = np.random.default_rng(42)
bootstrap_mae = np.array([
    np.mean(np.abs((y_pred - y_test)[idx]))
    for idx in (bootstrap_rng.integers(0, len(y_test), len(y_test)) for _ in range(2000))
])
mae_ci = np.quantile(bootstrap_mae, [0.025, 0.975])
print(f"IC bootstrap 95 % del MAE: [{mae_ci[0]:.3f}, {mae_ci[1]:.3f}]")
```

```python
from io import BytesIO
import matplotlib.pyplot as plt
from IPython.display import Image, display

fig, ax = plt.subplots(figsize=(6, 6))
ax.scatter(y_test, y_pred, alpha=0.75)
limits = [min(float(y_test.min()), float(y_pred.min())), max(float(y_test.max()), float(y_pred.max()))]
ax.plot(limits, limits, 'k--', linewidth=1)
ax.set_xlabel('KF observado')
ax.set_ylabel('KF predicho')
ax.set_title('EJECUCION_1: observado vs. predicho')
ax.grid(True, alpha=0.3)
fig.tight_layout()
png = BytesIO()
fig.savefig(png, format='png', dpi=120)
display(Image(data=png.getvalue()))
plt.close(fig)
```

## Interpretación y siguiente paso

Las métricas son para el conjunto de prueba separado aleatoriamente y no sustituyen una validación física independiente. Revisa que la expresión obtenida tenga sentido estructural, mantenga predicciones positivas en el rango de diseño y no dependa de extrapolaciones. Esta ejecución es solo la prueba inicial en `EJECUCION_1`; no entrena ni modifica los 1.356 casos de `RUN_2`.

```python
from itertools import product
import json

OUTPUT_DIR.mkdir(parents=True, exist_ok=True)
test_output = df.iloc[np.arange(len(df))].copy()
test_output['KF_predicho_test'] = np.nan
test_indices = df.index
_, test_rows = train_test_split(test_indices.to_numpy(), test_size=0.20, random_state=42)
test_output.loc[test_rows, 'KF_predicho_test'] = y_pred
test_output.to_csv(OUTPUT_DIR / 'predicciones_test.csv', index=False)

production_model = fit_symbolic(best_config, X, y, 'e1_modelo_final_todos_datos')
production_equation = production_model.sympy()
production_named = production_equation.xreplace({
    symbol: sp.Symbol(FEATURES[int(str(symbol)[1:])].upper())
    for symbol in production_equation.free_symbols
})
unique_values = [sorted(df[column].unique().tolist()) for column in FEATURES]
combination_count = int(np.prod([len(values) for values in unique_values]))
if combination_count > 100000:
    raise ValueError(f'El producto cartesiano ({combination_count}) supera el límite seguro de 100.000 combinaciones.')
combinations = pd.DataFrame(product(*unique_values), columns=FEATURES)
combinations['KF_predicho'] = production_model.predict(combinations[FEATURES].to_numpy(dtype=np.float64))
combinations.to_csv(OUTPUT_DIR / 'predicciones_combinaciones.csv', index=False)

validation_metrics.to_csv(OUTPUT_DIR / 'comparacion_configuraciones.csv', index=False)
pd.DataFrame([test_metrics | {
    'MAE_bootstrap_95_low': float(mae_ci[0]),
    'MAE_bootstrap_95_high': float(mae_ci[1]),
    'selected_configuration': best_name,
    'n_train': len(y_train),
    'n_test': len(y_test),
}]).to_csv(OUTPUT_DIR / 'metricas_test.csv', index=False)
equation_payload = {
    'worksheet': SHEET,
    'cases': len(df),
    'features': FEATURES,
    'target': 'KF',
    'KF_definition': '12*(1-nu^2)*FCR_24*hw/(pi^2*E*tw^3)',
    'E_N_per_mm2': E_N_PER_MM2,
    'nu': NU,
    'selected_configuration': best_name,
    'test_metrics': test_metrics,
    'MAE_bootstrap_95_interval': [float(mae_ci[0]), float(mae_ci[1])],
    'test_equation_sympy': str(best_equation),
    'final_equation_sympy': str(production_equation),
    'final_equation_named': str(production_named),
    'feature_order_for_x_symbols': ['SS_HW', 'A_R', 'DO_HW', 'HW_TW', 'N_PERF'],
    'combinations_predicted': combination_count,
}
(OUTPUT_DIR / 'expresion_y_validacion.json').write_text(
    json.dumps(equation_payload, indent=2, ensure_ascii=False), encoding='utf-8'
)
print('Resultados guardados en:', OUTPUT_DIR.resolve())
print('Archivos: expresion_y_validacion.json, metricas_test.csv, comparacion_configuraciones.csv, predicciones_test.csv, predicciones_combinaciones.csv')
print('Ecuación final ajustada a todos los casos:', production_named)
print('Combinaciones geométricas calculadas:', combination_count)
```
